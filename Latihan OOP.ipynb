```python
class Node():
  def __init__(self, data):
    self.data1 = data
    self.next = None

class LinkedList():
  def __init__(self):
    self.head = None

  def is_empty(self):
    return self.head is None

  def tambah_di_awal(self, data):
    new_node = Node(data)
    new_node.next = self.head
    self.head = new_node

  def __iter__(self):
    node = self.head
    while node is not None:
      yield node
      node = node.next

  def tambah_di_akhir(self, data):
    new_node = Node(data)
    if self.head is None:
      self.head = new_node
    else:
      current_node = self.head
      while current_node.next is not None:
        current_node = current_node.next
      current_node.next = new_node

  def tampilkan_kereta(self):
    current_node = self.head
    while current_node is not None:
      print(current_node.data1, end=" => ")
      current_node = current_node.next
    print("None")

kereta = LinkedList()
kereta.tambah_di_awal("Gerbong A")
kereta.tambah_di_awal("Gerbong B")
kereta.tambah_di_awal("Gerbong C")
kereta.tambah_di_akhir("Gerbong akhir")
kereta.tampilkan_kereta()
```

You can use `append()` to 'push' elements onto a list, which adds them to the end of the list. This is the standard way to add elements to a list, especially when treating it like a stack (Last-In, First-Out).

```python
my_stack = []
print(f"Initial stack: {my_stack}")

# Push elements using append()
my_stack.append("item1") # Push 'item1'
print(f"After pushing 'item1': {my_stack}")

my_stack.append("item2") # Push 'item2'
print(f"After pushing 'item2': {my_stack}")

my_stack.append("item3") # Push 'item3'
print(f"After pushing 'item3': {my_stack}")

# To 'pop' (remove the last item), you would use pop()
popped_item = my_stack.pop()
print(f"Popped item: {popped_item}, Stack after pop: {my_stack}")

popped_item = my_stack.pop()
print(f"Popped item: {popped_item}, Stack after pop: {my_stack}")
```

```python
class Hewan:
  def __init__(self, name):
    self.name = name

  def makan(self):
    print(f"{self.name} sedang makan segalanya")

class Kucing(Hewan):
  def __init__(self, name, warna_bulu, umur):
    super().__init__(name)
    self.warna_bulu = warna_bulu
    self.__umur = umur

  def makan(self):
    print(f"{self.name} berbulu {self.warna_bulu} sedang makan ikan dan sekarang berumur {self.__umur}")

  def getter (self):
    return self.__umur

  def setter (self, umur_baru):
    if umur_baru > 0:
      self.__umur = umur_baru
      print("umur berhasil ditambahkan")
    else:
      print("umur harus lebih dari 0")

hewan_kaki = Hewan("Hewan Kaki 4")
hewan_kaki.makan()

#objek children
hewan_kucing = Kucing("Kucing", "Calico", "32 Months")
hewan_kucing.makan()
hewan_kucing.setter(2)
```
