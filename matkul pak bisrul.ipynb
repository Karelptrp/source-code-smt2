```python
class Node():
  def __init__(self, data):
    self.data1 = data
    self.next = None

class LinkedList():
  def __init__(self):
    self.head = None

  def is_empty(self):
    return self.head is None

  def tambah_di_awal(self, data):
    new_node = Node(data)
    new_node.next = self.head
    self.head = new_node

  def __iter__(self):
    node = self.head
    while node is not None:
      yield node
      node = node.next

  def tambah_di_akhir(self, data):
    new_node = Node(data)
    if self.head is None:
      self.head = new_node
    else:
      current_node = self.head
      while current_node.next is not None:
        current_node = current_node.next
      current_node.next = new_node

  def tampilkan_kereta(self):
    current_node = self.head
    while current_node is not None:
      print(current_node.data1, end=" => ")
      current_node = current_node.next
    print("None")

kereta = LinkedList()
kereta.tambah_di_awal("Gerbong A")
kereta.tambah_di_awal("Gerbong B")
kereta.tambah_di_awal("Gerbong C")
kereta.tambah_di_akhir("Gerbong akhir")
kereta.tampilkan_kereta()
```

```python
my_stack = []
print(f"Initial stack: {my_stack}")

# Push elements using append()
my_stack.append("item1") # Push 'item1'
print(f"After pushing 'item1': {my_stack}")

my_stack.append("item2") # Push 'item2'
print(f"After pushing 'item2': {my_stack}")

my_stack.append("item3") # Push 'item3'
print(f"After pushing 'item3': {my_stack}")

# To 'pop' (remove the last item), you would use pop()
popped_item = my_stack.pop()
print(f"Popped item: {popped_item}, Stack after pop: {my_stack}")

popped_item = my_stack.pop()
print(f"Popped item: {popped_item}, Stack after pop: {my_stack}")
```

```python
import numpy as np

HistoryStack = np.array([], dtype=str)
RedoStack = np.array([], dtype=str)

def cetak_layar():
    print(f"Kalimat : {' '.join(HistoryStack)}")
    print(f"History : {HistoryStack}")
    print(f"Redo    : {RedoStack}")
    print("----------------------------------------")

def ketik(kata):
    global HistoryStack, RedoStack
    print(f"[Aksi] Ketik: '{kata}'")

    HistoryStack = np.append(HistoryStack, kata)


    RedoStack = np.array([], dtype=str)

    cetak_layar()

def undo():
    global HistoryStack, RedoStack
    print("[Aksi] Undo")

    if len(HistoryStack) > 0:
        data_pop = HistoryStack[-1]
        HistoryStack = HistoryStack[:-1]

        RedoStack = np.append(RedoStack, data_pop)

    cetak_layar()

def redo():
    global HistoryStack, RedoStack
    print("[Aksi] Redo")

    if len(RedoStack) > 0:
        data_pop = RedoStack[-1]
        RedoStack = RedoStack[:-1]

        HistoryStack = np.append(HistoryStack, data_pop)

    cetak_layar()

ketik("Halo")
ketik("Dunia")
undo()
redo()
```

```python
class textnambahkurang:
    def __init__(self):
        self.history_stack = []  #HistoryStack
        self.redo_stack = []     #RedoStack
        self.current_text = ""

    def ketik(self, teks):
        #Simpan kondisi sebelumnya ke history
        self.history_stack.append(self.current_text)
        #Tambah teks baru
        self.current_text += teks
        #Kosongkan redo stack
        self.redo_stack.clear()
        self.tampilkan()

    def undo(self):
        if self.history_stack:
            #Pindahkan ke redo
            self.redo_stack.append(self.current_text)
            #Ambil dari history
            self.current_text = self.history_stack.pop()
        else:
            print("ga bisa diundo")
        self.tampilkan()

    def redo(self):
        if self.redo_stack:
            #Pindahkan ke history
            self.history_stack.append(self.current_text)
            #Ambil dari redo
            self.current_text = self.redo_stack.pop()
        else:
            print("ga bisa diredo")
        self.tampilkan()

    def tampilkan(self):
        print("pertanyaan:", self.current_text)


# Contoh penggunaan
nana = textnambahkurang()

nana.ketik("nama")
print(nana.history_stack)
print(nana.redo_stack)
nana.ketik("lo")
print(nana.history_stack)
print(nana.redo_stack)
nana.undo()
print(nana.history_stack)
print(nana.redo_stack)
nana.undo()
print(nana.history_stack)
print(nana.redo_stack)
nana.redo()
print(nana.history_stack)
print(nana.redo_stack)
nana.ketik("siapa?")
print(nana.history_stack)
print(nana.redo_stack)
nana.redo()
print(nana.history_stack)
print(nana.redo_stack)  #ini tidak akan jalan karena redo sudah di-reset
```

```python
import numpy as np

class MesinWaktu:
    def __init__(self):
        self.history_stack = np.array([])
        self.redo_stack = np.array([])

    def _cetak_layar(self):
        print(f"Kalimat : {' '.join(self.history_stack)}")
        print(f"History : {self.history_stack}")
        print(f"Redo    : {self.redo_stack}")
        print("----------------------------------------")

    def ketik(self, kata):
        print(f"[Aksi] Ketik: '{kata}'")
        self.history_stack = np.append(self.history_stack, kata)
        self.redo_stack = np.array([])
        self._cetak_layar()

    def undo(self):
        print("[Aksi] Undo")
        if len(self.history_stack) > 0:
            data_pop = self.history_stack[-1]
            self.history_stack = self.history_stack[:-1]
            self.redo_stack = np.append(self.redo_stack, data_pop)
        self._cetak_layar()

    def redo(self):
        print("[Aksi] Redo")
        if len(self.redo_stack) > 0:
            data_pop = self.redo_stack[-1]
            self.redo_stack = self.redo_stack[:-1]
            self.history_stack = np.append(self.history_stack, data_pop)
        self._cetak_layar()

editor = MesinWaktu()
editor.ketik("Halo")
editor.ketik("Dunia")
editor.undo()
editor.redo()
editor.ketik("Python")
editor.undo()
editor.ketik("Colab")
```

### Contoh Penggunaan `scipy.stats`

Modul `scipy.stats` menyediakan berbagai fungsi untuk analisis statistik, termasuk distribusi probabilitas, fungsi statistik, dan uji statistik. Berikut adalah contoh sederhana untuk menghitung statistik deskriptif dari sebuah data.

```python
import numpy as np
from scipy import stats

# Membuat data sampel acak
data_sample = np.random.normal(loc=50, scale=10, size=100)

print("Data Sampel:", data_sample[:10]) # Tampilkan 10 data pertama
print("-------------------------------------")

# Menghitung statistik deskriptif menggunakan scipy.stats

# Mean (Rata-rata)
mean_value = np.mean(data_sample) # Menggunakan numpy untuk mean, bisa juga stats.tmean
# stats.tmean(data_sample) bisa digunakan juga
print(f"Mean (Rata-rata): {mean_value:.2f}")

# Median (Nilai tengah)
median_value = np.median(data_sample) # Menggunakan numpy untuk median, bisa juga stats.tmedian
print(f"Median (Nilai tengah): {median_value:.2f}")

# Mode (Nilai yang paling sering muncul)
# mode_result = stats.mode(data_sample)
# print(f"Mode (Nilai paling sering muncul): {mode_result.mode[0]:.2f} (Count: {mode_result.count[0]})")

# Variance (Variansi)
variance_value = np.var(data_sample) # Menggunakan numpy untuk variance, bisa juga stats.tvar
print(f"Variance (Variansi): {variance_value:.2f}")

# Standard Deviation (Standar Deviasi)
std_dev_value = np.std(data_sample) # Menggunakan numpy untuk standar deviasi, bisa juga stats.tstd
print(f"Standard Deviation (Standar Deviasi): {std_dev_value:.2f}")


# Contoh lain: Menghitung skewness dan kurtosis
skewness_value = stats.skew(data_sample)
kurtosis_value = stats.kurtosis(data_sample)
print(f"Skewness: {skewness_value:.2f}")
print(f"Kurtosis: {kurtosis_value:.2f}")
```

# stats.norm(loc, scale)

scipy.stats.norm(loc, scale) mendefinisikan variabel acak kontinu normal (Gaussian) di Python, di mana loc menetapkan nilai rata-rata (\(\mu \)) dan scale menetapkan simpangan baku (\(\sigma \)). Fungsi ini bertindak sebagai objek distribusi tetap, secara default menggunakan distribusi normal standar (\(\mu=0, \sigma=1\)) jika parameter dihilangkan.

```python
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt

print("### Penjelasan `scipy.stats.norm(loc, scale)`")
print("--------------------------------------------------")
print("`scipy.stats.norm` adalah objek distribusi normal dari modul statistik SciPy. Ini memungkinkan kita untuk bekerja dengan distribusi normal, seperti menghitung PDF (Probability Density Function), CDF (Cumulative Distribution Function), dan menghasilkan angka acak.")

print("\n### Parameter:")
print("1. `loc` (location): Ini adalah parameter rata-rata (mean) dari distribusi normal. Ini menentukan di mana puncak kurva distribusi berada.")
print("2. `scale` (scale): Ini adalah parameter standar deviasi dari distribusi normal. Ini menentukan seberapa 'tersebar' kurva distribusi itu; nilai `scale` yang lebih besar berarti distribusi yang lebih lebar.")

print("\n### Contoh Penggunaan:")

# Membuat objek distribusi normal
# Rata-rata (mean) = 0, Standar Deviasi (std dev) = 1 (distribusi normal standar)
distribusi_normal_standar = stats.norm(loc=0, scale=1)
print(f"Distribusi Normal Standar (loc=0, scale=1): {distribusi_normal_standar}")

# Membuat objek distribusi normal dengan rata-rata 50 dan standar deviasi 10
distribusi_kustom = stats.norm(loc=50, scale=10)
print(f"Distribusi Kustom (loc=50, scale=10): {distribusi_kustom}")

print("\n--- Fungsi-fungsi Umum --- ")
# 1. Menghitung Probability Density Function (PDF) pada suatu titik
x = 1
pdf_value = distribusi_normal_standar.pdf(x)
print(f"PDF dari distribusi normal standar di x={x}: {pdf_value:.4f}")

x_custom = 55
pdf_value_custom = distribusi_kustom.pdf(x_custom)
print(f"PDF dari distribusi kustom di x={x_custom}: {pdf_value_custom:.4f}")

# 2. Menghasilkan angka acak dari distribusi
sampel_acak = distribusi_kustom.rvs(size=5)
print(f"\n5 sampel acak dari distribusi kustom: {sampel_acak}")

# 3. Menghitung Cumulative Distribution Function (CDF) pada suatu titik
cdf_value = distribusi_normal_standar.cdf(x)
print(f"\nCDF dari distribusi normal standar di x={x}: {cdf_value:.4f}")

# Visualisasi (opsional, membutuhkan matplotlib)
print("\nVisualisasi PDF (membutuhkan matplotlib):")
fig, ax = plt.subplots(1, 1)
x_values = np.linspace(distribusi_kustom.ppf(0.001), distribusi_kustom.ppf(0.999), 100)
ax.plot(x_values, distribusi_kustom.pdf(x_values), 'r-', lw=2, alpha=0.6, label='pdf normal kustom')
ax.plot(x_values, distribusi_normal_standar.pdf(x_values), 'b-', lw=2, alpha=0.6, label='pdf normal standar')
ax.legend(loc='best', frameon=False)
plt.title('Probability Density Function (PDF) Normal Distribution')
plt.xlabel('Nilai')
plt.ylabel('Probabilitas')
plt.grid(True)
plt.show()
```

```python
class BatchData:
  def __new__ (cls, *args):
    print(f"Mengalokasikan memori untuk {cls.__name__} menggunakan __new")
    return super().__new__(cls)
  def __init__(self, nama_batch, data_list):
    print(f"Menginisiasi untuk data {nama_batch} menggunakan __init__")
    self.nama_batch = nama_batch
    self.data_list = data_list
  def __str__(self):
    return f"Batch {self.nama_batch} memiliki {len(self)} baris data"
  def __len__(self):
    return len(self.data_list)
  def __add__(self, batch_lain):
    print(f"Proses Pengembangan {self.nama_batch} dengan {batch_lain.nama_batch}")
    nama_batch_add = f"{self.nama_batch} + {batch_lain.nama_batch}"
    data_list_add = self.data_list + batch_lain.data_list
    return nama_batch_add, data_list_add



print("== Pembuatan Dataset 1 ==")
batch1 = BatchData("Data_April", [1,2,3,4,5])
print("== Pembuatan Dataset 2 ==")
batch2 = BatchData("Data_Mei", [6,7,8,9,10])
print("== Rekap Data ==")
print(batch1)
print(batch2)
print("== Batch Total ==")
batch_total = batch1 + batch2
print(batch_total)
```

```python
import time
#menggunakan decorator
def hitung_durasi(fungsi_asli):
  def pembungkus(*args, **kwargs):

#logika awal
    start = time.time()

  #logika inti pelatihan
    hasil = fungsi_asli(*args, **kwargs)

  #logika loging
    end = time.time()
    durasi = end - start
    print(f"Proses training model membutuhkan waktu {durasi} detik")
    return hasil
  return pembungkus
```

```python
import time
#menggunakan decorator
def hitung_durasi(fungsi_asli):
  def pembungkus(*args, **kwargs):

#logika awal
    start = time.time()

  #logika inti pelatihan
    hasil = fungsi_asli(*args, **kwargs)

  #logika loging
    end = time.time()
    durasi = end - start
    print(f"Proses training model membutuhkan waktu {durasi} detik")
    return hasil
  return pembungkus

@hitung_durasi
def latih_model(nama_model, jumlah_data):
  print(f" Memulai proses training model: {nama_model}...")
  print (f" Membaca {jumlah_data} baris data...")
  time.sleep(2)

if __name__ == "__main__":
  print("== SIMULASI SISTEM ==")
  # Menjalankan fungsi yang sudah dibungkus decorator
  hasil_sistem = latih_model("Random Forest", jumlah_data=50000)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    'Tanggal': pd.date_range(start='2026-05-01', periods=14, freq='D'),
    'Jenis_Konten': ['Reels', 'Carousel', 'Feed', 'Reels', 'Carousel', 'Feed', 'Reels'] *2,
    'Reach': [2000, 1500, 1000,2200,1600,1100,2500,2100,1400,950,2300,1700,1050,2400],
    'Pendaftaran': [20,15,10,22,16,11,25,21,14,9,23,17,10,24]
}

df = pd.DataFrame(data)

print("TAMPILAN DATA MENTAH (5 BARIS PERTAMA)")
display(df.head())
```

```python
rata_reach = df.groupby('Jenis_Konten')['Reach'].mean()
print("Rata-rata Reach berdasarkan Jenis Konten:")
display(rata_reach)
```

```python
total_reach = df.groupby('Jenis_Konten')['Reach'].sum()
print("Total Reach dan Pendaftar berdasarkan Jenis Konten:")
display(total_reach)


total_daftar = df.groupby('Jenis_Konten')['Pendaftaran'].sum()
display(total_daftar)


ringkasan = df.groupby('Jenis_Konten').agg(
    rata_reach=('Reach', 'mean'),
    total_reach=('Reach', 'sum'),
    total_daftar=('Pendaftaran', 'sum')
).reset_index()


print("Ringkasan Data")
display(ringkasan)
```
